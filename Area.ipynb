## THIS PROGRAM CALCULATES THE AREA AND PERIMETER OF A SQUARE

```python
length= float(input("Enter the length of the square in cm: "))
## typecasts the input to a float

area= length ** 2 
## uses the ** operator to calculate the area of the square


perimeter= length * 4
## uses the * operator to calculate the perimeter of the square
print("The area of the square is", area)
print("The perimeter of the square is", perimeter)
## PRINT STATEMENTS
```
